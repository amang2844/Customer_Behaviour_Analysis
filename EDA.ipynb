```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#crerate a column
labels=['Young Adult','Adult','Middle-aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(6)
```

```python
#create column purchase_frequency_days 

frequency_mapping={
    'Fortnightly':14,
    'Weekly':7, 
    'Monthly':30, 
    'Quarterly':90, 
    'Bi-Weekly':14, 
    'Annually':365, 
    'Every 3 Months':90  
}

df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```

```python

from sqlalchemy import create_engine
import urllib

# 1. Define your server and database
server = 'DESKTOP-IEJTJGP\SQLEXPRESS'
database = 'customer_behavior'
driver = 'ODBC Driver 17 for SQL Server' # Often this is the correct driver

# 2. Create the encoded connection string for Windows Authentication
# This line makes the 'address' readable for the driver
params = urllib.parse.quote_plus(f"DRIVER={{{driver}}};SERVER={server};DATABASE={database};Trusted_Connection=yes;")

# 3. Create the database engine (The Delivery Company)
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")
```

```python
# Assuming your data is ready in a DataFrame called 'df'
table_name = 'My_Analysis_Data_Table'

# Use the Pandas 'to_sql' method to push the data!
df.to_sql(
    name=table_name, 
    con=engine, 
    if_exists='replace', # You can use 'append' if the table exists
    index=False
)

print(f"Success! Data loaded to SQL Server table: {table_name}")
```
